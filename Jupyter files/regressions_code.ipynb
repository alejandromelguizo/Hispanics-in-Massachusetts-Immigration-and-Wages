```python
#regression on US hispanic data
model = smf.ols(formula = """
log_wage ~ 
AGE +
years_since_immig + 
sex_clean + 
C(EMPSTAT) + 
US_educ +
sex_clean * C(educ_clean) + 
sex_clean * C(hisp_clean) +
C(CITIZEN) +
C(hisp_clean) + 
C(educ_clean) + 
C(race_clean) + 
C(STATEFIP)
"""
, data = df_hisp)
results = model.fit(cov_type='HC1') #HC1 adds robust SE 

print(results.summary())
```

## Different test versions of regressions:

```python
#adapting regression model from US to work on collinearity


#removed AGE to see how it affects multicollinearity
model = smf.ols(formula = """
log_wage ~ 
years_since_immig + 
sex_clean + 
C(EMPSTAT) + 
US_educ +
sex_clean * C(educ_clean) + 
sex_clean * C(hisp_clean) +
C(CITIZEN) +
C(hisp_clean) + 
C(educ_clean) + 
C(race_clean) + 
C(STATEFIP)
"""
, data = df_hisp)
results = model.fit(cov_type='HC1') #HC1 adds robust SE 

print(results.summary())
```

```python
#lower variable count
model = smf.ols(formula = """
log_wage ~ 
years_since_immig + 
sex_clean + 
US_educ +
C(hisp_clean) 
"""
, data = df_hisp)
results = model.fit(cov_type='HC1') #HC1 adds robust SE 

print(results.summary())
```

```python
#educ_clean rather than US_educ
model = smf.ols(formula = """
log_wage ~ 
years_since_immig + 
years_since_immig * years_since_immig +
sex_clean + 
C(educ_clean) +
C(hisp_clean) 
"""
, data = df_hisp)
results = model.fit(cov_type='HC1') #HC1 adds robust SE 

print(results.summary())
```
